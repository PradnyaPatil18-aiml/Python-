```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, 30, 35, 40, 28],
    'City': ['New York', 'London', 'Paris', 'New York', 'London'],
    'Salary': [70000, 80000, 120000, 90000, 75000]
}

df = pd.DataFrame(data)
print(df)

print(df.head(2))
print(df.info())
print(df.describe())

print(df['Name'])
print(df[['Name', 'Salary']])
print(df.loc[0])
print(df.iloc[1:4])

print(df[df['Age'] > 30])
print(df[df['City'] == 'London'])

df['Bonus'] = df['Salary'] * 0.10
print(df)

city_salary = df.groupby('City')['Salary'].mean()
print(city_salary)

sorted_df = df.sort_values(by='Age', ascending=False)
print(sorted_df)

df_with_nan = df.copy()
df_with_nan.loc[2, 'Salary'] = None

print(df_with_nan)

mean_salary = df_with_nan['Salary'].mean()
df_with_nan['Salary'] = df_with_nan['Salary'].fillna(mean_salary)
print(df_with_nan)
```

```python
import pandas as pd

data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard'],
    'Price': [1200, 25, 75],
    'Stock': [15, 100, 50]
}
df = pd.DataFrame(data)

# --- WRITING TO AN EXCEL FILE ---
# index=False excludes row numbers; sheet_name specifies the sheet tab name
df.to_excel('output.xlsx', sheet_name='Inventory', index=False)


# --- READING FROM AN EXCEL FILE ---
df_from_excel = pd.read_excel('output.xlsx', sheet_name='Inventory')
print(df_from_excel)
```
