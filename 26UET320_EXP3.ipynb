```python
username = "dkte_python"

if len(username) < 3:
    print("Error: Username must be at least 3 characters long.")
elif len(username) > 15:
    print("Error: Username cannot exceed 15 characters.")
elif " " in username:
    print("Error: Username cannot contain spaces.")
else:
    print(f"Success! '{username}' is a valid username.")
```
