```python
fruits = ["apple", "banana", "cherry"]
print("fruits:",fruits)
chars = list("Python")
print("result=",chars)
squares = [x**2 for x in range(5)]
print("result=",squares)
zeros = [0] * 5
print("result=",zeros)
```

```python
colors = ["red", "green", "blue"]

# append()
colors.append("yellow")
print(colors)  # ['red', 'green', 'blue', 'yellow']

# insert()
colors.insert(1, "purple")
print(colors)  # ['red', 'purple', 'green', 'blue', 'yellow']

# remove()
colors.remove("green")
print(colors)  # ['red', 'purple', 'blue', 'yellow']

# pop()
popped_item = colors.pop()
print(popped_item)  # 'yellow'

# sort
numbers = [4, 1, 8, 3]
numbers.sort()
print(numbers)  # [1, 3, 4, 8]

# reverse()
numbers.reverse()
print(numbers)
```

```python
scores = [85, 92, 78, 92, 60]

print(len(scores))
print(max(scores))
print(min(scores))
print(sum(scores))
print(sorted(scores))
```

```python
numbers = [10, 20, 30, 40, 50, 60, 70]

#  slicing
print(numbers[1:4])
print(numbers[:3])
print(numbers[3:])
print(numbers[::-1])
```

```python
evens = [x for x in range(10) if x % 2 == 0]
print(evens)

words = ["hello", "world", "python"]
upper_words = [w.upper() for w in words]
print(upper_words)  # ['HELLO', 'WORLD', 'PYTHON']
```
