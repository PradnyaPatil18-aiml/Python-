```python

fruits = {"apple", "banana", "cherry"}
print(fruits)

my_tuple = ("red", "green", "blue", "red")
color_set = set(my_tuple)
print(color_set)
empty_set = set()


squares = {x**2 for x in range(6)}
print(squares)
```

```python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# Union (|)
print(set_a | set_b)
#Intersection
print(set_a & set_b)

# Difference (-):
print(set_a - set_b)

# Symmetric Difference (^):
print(set_a ^ set_b)
```

```python
even_numbers = {2, 4, 6, 8}
odd_numbers = {1, 3, 5, 7}
prime_numbers = {2, 3, 5, 7}
print(even_numbers.isdisjoint(odd_numbers))
print(even_numbers.isdisjoint(prime_numbers))
```

```python
group_a = {1, 2}
group_b = {1, 2, 3, 4, 5}
group_c = {4, 6}

# Using issubset() method
print(group_a.issubset(group_b))
print(group_c.issubset(group_b))

# Using the operator shortcut (<=)
print(group_a <= group_b)
```

```python
# add(): Adds a single element to the set
my_set = {1, 2, 3}
my_set.add(4)
print(my_set)  # {1, 2, 3, 4}

# update(): Adds multiple elements from another iterable (like a tuple or list)
my_set.update((5, 6, 7))  # Passing a tuple
print(my_set)  # {1, 2, 3, 4, 5, 6, 7}

# remove(): Removes a specific element (raises an error if the element is not found)
my_set.remove(2)

# discard(): Removes a specific element (does NOT raise an error if not found)
my_set.discard(99)

# pop(): Removes and returns a random element from the set
removed_item = my_set.pop()
print(f"Popped item: {removed_item}")
```
