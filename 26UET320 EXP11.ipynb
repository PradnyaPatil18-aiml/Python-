```python
import matplotlib.pyplot as plt
hours =[4,5,6,7,8,9]
marks =[50,60,80,100,90,45]
plt.scatter(hours,marks,color="Blue",marker="o",s = 100,label ="students")
plt.title("Hours studied vs marks obtained")
plt.xlabel("Hours studied")
plt.ylabel("Marks obtained")
plt.grid(True)
plt.legend()
plt.show()
```

```python
hours =[4,5,6,7,8,9]
marks =[50,60,80,100,90,45]
plt.scatter(hours,marks,color="Blue",marker="o", s=120)
plt.title("Hours studied vs marks obtained")
plt.xlabel("Hours studied")
plt.ylabel("Marks obtained")
plt.plot(hours,marks,color="red",label ="lineplot")
plt.grid(True)
plt.legend()
plt.show()
```
